# Pydantic Nested Models

### 1. What are Nested Models?
- Nested models allow us to use one Pydantic model inside another model.
- They help represent complex and structured data.
- They are useful when data contains objects within objects.

### 3. Important Points
- `Address` is a separate Pydantic model.
- `address: Address` specifies that the `address` field must follow the `Address` schema.
- Pydantic validates the nested data automatically.
- Invalid nested data raises a `ValidationError`.
- Nested models make complex data easier to organize and maintain.

### Remember
**Nested Model = A Pydantic model used as a field inside another Pydantic model.**

```python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    pincode: int

class Student(BaseModel):
    name: str
    age: int
    address: Address  # Nested model

student = Student(
    name="Shivansh",
    age=21,
    address={
        "city": "Dehradun",
        "pincode": 248001
    }
)

print(student.address.city)
print(student.address.pincode)
```
