# Pydantic Serialization

### 1. What is Serialization?
- Serialization is the process of converting a Pydantic model object into a format that can be stored or transmitted, such as a dictionary or JSON.
- It is useful when sending data through APIs, storing data, or sharing information between applications.

### 2. Important Methods

- `model_dump()` → Converts a Pydantic object into a Python dictionary.
- `model_dump_json()` → Converts a Pydantic object into a JSON string.
- `model_dump(exclude=...)` → Excludes specified fields.
- `model_dump(include=...)` → Includes only specified fields.


### 4. Why is Serialization Useful?
- **APIs:** Send structured data between applications.
- **Storage:** Prepare data for saving.
- **Data Exchange:** Convert model objects into standard formats.
- **Filtering:** Include or exclude specific fields.

### Remember
**Serialization = Pydantic Object → Dictionary / JSON**

**Deserialization = Input Data → Pydantic Object**

```python
# ---------------- Pydantic Model we made in Basics-----------------------------

from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict, Optional, Annotated


# STEP 1: Define a Pydantic model and specify the schema
class Students(BaseModel):
    name: str = Field(max_length=50)
    email: EmailStr                 # EmailStr = Validates email format
    age: int = Field(gt=18, lt=60)  # Field = Add constraints and to attach MetaData,Default Values
    LinkedIn: AnyUrl                # AnyUrl = Validates URL format
    roll_no: int
    cgpa: float
    subjects: List[str]
    contact_details: Dict[str, str]
    marriage_status: Optional[bool] = None
    weight: Annotated[float,Field(
            gt=0,
            title="Body Weight",
            description="Weight of the product in kilograms. Must be greater than 0.",
            examples=[1.5]
        )
    ]     #we will see annotations in .json 


# STEP 2: Define a function that accepts a Students object
def insert_student_data(student: Students):
    print(f"Name: {student.name}")


# STEP 3: Create a dictionary containing raw student data
student1_info = {
    'name': 'Anisha Hazra',
    'email': 'anishahazra1201@gmail.com',
    'age': 21,
    'LinkedIn': 'https://www.linkedin.com/in/username/',
    'roll_no': 169,
    'cgpa': 7.8,
    'subjects': ['Maths', 'Science', 'Graphic Design'],
    'contact_details': {
        'city': 'Navi Mumbai',
        'Phone_number': '+91 9167395315'
    },
    'weight': 66
}


# STEP 4: Unpack the dictionary and create a Pydantic object
# Pydantic validates the data against the Students schema
student1 = Students(**student1_info)


# STEP 5: Pass the validated Students object to the function
insert_student_data(student1)





#Serialisation



temp= student1.model_dump(include=['name','age'])
print(temp)
print(type(temp))
```

```python
from pydantic import BaseModel

# Definition: exclude_unset=True excludes fields that were not
# explicitly provided when creating the Pydantic object.

class Student(BaseModel):
    name: str
    age: int = 18
    city: str = "Dehradun"

student = Student(name="Shivansh")

# Exclude fields that were not explicitly provided
print(student.model_dump(exclude_unset=True))

# Normal serialization includes default values
print(student.model_dump())
```
