# Pydantic Model Validator

### 1. What is a Model Validator?

- `@model_validator` is used to validate the entire model rather than an individual field.
- It is useful when validation depends on multiple fields.
- It is imported from `pydantic`.

### 2. Important Points

- `@model_validator` → Validates the model as a whole.
- `mode="after"` → Runs after the model fields have been validated.
- `self` → Provides access to the model's fields.
- `raise ValueError()` → Rejects invalid data.
- `return self` → Returns the validated model instance.

### 3. Validation Modes

- **Before:** Validates raw input before model validation.
- **After:** Validates the model after its fields have been validated.
- **Wrap:** Allows control over the model validation process.

### 4. Field Validator vs. Model Validator

- `@field_validator` → Validates an individual field.
- `@model_validator` → Validates relationships between multiple fields.

### 5. Example: Patient Emergency Contact

**Scenario:** If a patient's age is above 60, an emergency contact number must be provided. Otherwise, the patient object cannot be created.

This is an example of **cross-field validation**, because the validation of the emergency contact depends on the patient's age.

### Remember

**Field Validator = Individual Field**

**Model Validator = Entire Model / Relationships Between Fields**

```python
from pydantic import BaseModel, model_validator
from typing import List, Dict, Optional


# STEP 1: Define a Pydantic model and specify the schema
class Students(BaseModel):
    name: str
    email: str                 
    age: int
    subjects: List[str]               # List of strings
    contact_details: Dict[str, str]   # Dictionary with string keys and values
    marriage_status: Optional[bool] = None
    weight: float 


    @model_validator(mode='after')
    def Validate_Emergency_Contact(cls, model):
        if model.age<10 and 'emergency' not in model.contact_details:
            raise ValueError("Students younger than age 10 must have to fill emergency contact") 
        return model   



# STEP 2: Define a function that accepts a Students object
def insert_student_data(student: Students):
    print(f"Name: {student.name}")
    print(f"Email: {student.email}")
    print(f"Age: {student.age}")
    print(f"Subjects: {student.subjects}")
    print(f"Contact Details: {student.contact_details}")
    print(f"Marriage Status: {student.marriage_status}")
    print(f"Weight: {student.weight}")


# STEP 3: Create a dictionary containing raw student data
student1_info = {
    'name': 'Baccha',
    'email': '',
    'age': 12,
    'subjects': ['Maths', 'Science', 'Graphic Design'],
    'contact_details': {
        'city': 'Navi Mumbai',
        'Phone_number': '+91 9167395315'
    },
    'weight': 12.5
}


# STEP 4: Create a Pydantic object and validate the data
student1 = Students(**student1_info)


# STEP 5: Pass the validated object to the function
insert_student_data(student1)
```
