# Pydantic Field Validator

### What is a Field Validator?
- `@field_validator` is used to apply custom data validation rules to a specific field in a Pydantic model.
- It checks whether a field's value satisfies our requirements.
- If validation fails, it raises a `ValidationError`.

### Important Points
- `@field_validator("age")` → Specifies the field to validate.
- `@classmethod` → Defines the validator as a class method.
- `cls` → Refers to the model class.
- `value` → Contains the field's value.
- `raise ValueError()` → Rejects invalid data.
- `return value` → Returns the validated value.

### 4. Validation Modes
- **Before:** Validates or transforms raw input before standard validation.for ex:- age is in string it will pass as a string
- **After(Default):** Validates the value after standard type validation; this is the default. age will first convert in int then pass to the field-validator
- **Plain:** Replaces standard validation for that field.
- **Wrap:** Allows control over standard validation.

### 5. Field() vs. @field_validator
- `Field()` → Defines built-in constraints and metadata, such as `gt=0`.
- `@field_validator` → Implements custom validation logic.

### Remember
**Field() = Built-in rules | @field_validator = Custom rules**

```python
from pydantic import BaseModel, EmailStr, Field, field_validator
from typing import List, Dict, Optional


# STEP 1: Define a Pydantic model and specify the schema
class Students(BaseModel):
    name: str
    email: EmailStr                  # Validates email format
    age: int
    subjects: List[str]               # List of strings
    contact_details: Dict[str, str]   # Dictionary with string keys and values
    marriage_status: Optional[bool] = None
    weight: float = Field(gt=0)       # Weight must be greater than 0

    # Custom field validator for email domains
    @field_validator('email')
    @classmethod
    def email_validator(cls, value):
        valid_domains = ['hdfc.com', 'icici.com', 'axis.com']

        domain_name = str(value).split('@')[-1].lower()

        if domain_name not in valid_domains:
            raise ValueError("Email must belong to hdfc.com, icici.com, or axis.com")

        return value

    @field_validator('name')
    @classmethod
    def transform_name(cls, value):
        return value.upper()


    @field_validator('age', mode='before')
    @classmethod
    def age_validator(cls, value):
        if value>21 and value<60:
            return value
        else:
            raise ValueError("Age should be in between 21 and 60")
    



# STEP 2: Define a function that accepts a Students object
def insert_student_data(student: Students):
    print(f"Name: {student.name}")
    print(f"Email: {student.email}")
    print(f"Age: {student.age}")
    print(f"Subjects: {student.subjects}")
    print(f"Contact Details: {student.contact_details}")
    print(f"Marriage Status: {student.marriage_status}")
    print(f"Weight: {student.weight}")


# STEP 3: Create a dictionary containing raw student data
student1_info = {
    'name': 'Anisha Hazra',
    'email': 'anisha@hdfc.com',
    'age': 22,
    'subjects': ['Maths', 'Science', 'Graphic Design'],
    'contact_details': {
        'city': 'Navi Mumbai',
        'Phone_number': '+91 9167395315'
    },
    'weight': 64.2
}


# STEP 4: Create a Pydantic object and validate the data
student1 = Students(**student1_info)


# STEP 5: Pass the validated object to the function
insert_student_data(student1)
```
