# Pydantic Computed Fields

### 1. What is a Computed Field?
- `@computed_field` is used to include a property calculated from other fields in a Pydantic model.
- It avoids storing redundant data because the value is calculated automatically.
- It is commonly used with Python's `@property` decorator.

### 3. Important Points
- `@computed_field` → Includes a calculated property in serialization.
- `@property` → Calculates the value when accessed.
- Computed fields are derived from other fields.
- They are included in `model_dump()` by default.
- They are not normally accepted as input fields during model creation.

### Remember
**Computed Field = Automatically Calculated Value Based on Other Fields**

```python
from pydantic import BaseModel, computed_field
from typing import List, Dict, Optional


# STEP 1: Define a Pydantic model and specify the schema
class Students(BaseModel):
    name: str
    email: str
    age: int
    subjects: List[str]
    contact_details: Dict[str, str]
    marriage_status: Optional[bool] = None
    weight: float  # Weight in kg
    height: float  # Height in meters

    # Computed field: Automatically calculates BMI
    @computed_field
    @property
    def BMI(self) -> float:
        return round(self.weight / (self.height ** 2), 2)


# STEP 2: Define a function that accepts a Students object
def insert_student_data(student: Students):
    print(f"Name: {student.name}")
    print(f"Email: {student.email}")
    print(f"Age: {student.age}")
    print(f"Subjects: {student.subjects}")
    print(f"Contact Details: {student.contact_details}")
    print(f"Marriage Status: {student.marriage_status}")
    print(f"Weight: {student.weight} kg")
    print(f"Height: {student.height} m")
    print(f"BMI: {student.BMI}")


# STEP 3: Create a dictionary containing raw student data
student1_info = {
    'name': 'Baccha',
    'email': 'baccha@example.com',
    'age': 12,
    'subjects': ['Maths', 'Science', 'Graphic Design'],
    'contact_details': {
        'city': 'Navi Mumbai',
        'Phone_number': '+91 9167395315'
    },
    'weight': 77,
    'height': 1.73
}


# STEP 4: Create a Pydantic object and validate the data
student1 = Students(**student1_info)


# STEP 5: Pass the validated object to the function
insert_student_data(student1)
```
