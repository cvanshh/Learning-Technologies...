# Pydantic

## What is Pydantic?

**Pydantic is a Python library used to validate data and ensure it follows a predefined structure, data types, and validation rules.**

Python is a dynamically typed language, meaning a variable can hold values of different data types at different times. This flexibility can sometimes lead to inconsistent data, making it difficult to process and store reliably.

Therefore, programmers need to define a clear **schema** that specifies the expected structure and data types. Pydantic helps achieve this by validating data against predefined rules.

### Example

```python
age = 30       # Integer
age = "thirty" # String
```

Python allows both assignments, but if an application expects `age` to be an integer, the string `"thirty"` may cause problems.

**Pydantic reduces repetitive manual validation by allowing developers to define a schema once and validate data against predefined rules.**

---

# Pydantic — 3-Step Workflow

## 1. Define a Pydantic Model (Class)

- Create a class that inherits from `BaseModel`.
- Define the expected fields and their data types.
- Add validation constraints wherever required.
- Example: `Field(gt=0)` means the value must be greater than zero.

## 2. Instantiate the Model (Create an Object)

- Pass raw input data, usually a dictionary or JSON-like data, into the model.
- Pydantic automatically validates the input.
- It converts compatible data types when possible.
- If the data violates the model's rules or has an invalid type, Pydantic raises a `ValidationError`.

## 3. Use the Validated Object

- If validation succeeds, Pydantic returns a model object.
- Access its fields using dot notation.
- Pass this object to functions or use it throughout your application.

---

## Example

```python
from pydantic import BaseModel, Field

# Step 1: Define the Pydantic model and schema
class Student(BaseModel):
    name: str
    age: int = Field(gt=0)

# Step 2: Create an object and validate the data
student = Student(name="Shivansh", age="21")

# Step 3: Access the validated data
print(student.name)  # Shivansh
print(student.age)   # 21
```

### Explanation

- `BaseModel`: Base class used to create Pydantic models.
- `name: str`: Name must be a string.
- `age: int`: Age must be an integer.
- `Field(gt=0)`: Age must be greater than zero.
- `Student(...)`: Creates a Pydantic object and validates the input.
- `"21"` is converted to `21` because Pydantic can convert this compatible input.

---

## Benefits of Pydantic

- **Structured Data:** Defines a clear schema for data.
- **Automatic Validation:** Checks input against predefined rules.
- **Type Conversion:** Converts compatible input types when possible.
- **Error Detection:** Raises `ValidationError` when validation fails.
- **Less Manual Code:** Reduces repetitive validation logic.
- **Improved Reliability:** Helps maintain consistent data throughout an application.

## Remember

**Define Class → Pass Raw Data → Validate → Get Pydantic Object → Use Data**

````python
# ---------------- Pydantic for Type Validation ----------------

from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict, Optional, Annotated


# STEP 1: Define a Pydantic model and specify the schema
class Students(BaseModel):
    name: str = Field(max_length=50)
    email: EmailStr                 # EmailStr = Validates email format
    age: int = Field(gt=18, lt=60)  # Field = Add constraints and to attach MetaData,Default Values
    LinkedIn: AnyUrl                # AnyUrl = Validates URL format
    roll_no: int
    cgpa: float
    subjects: List[str]
    contact_details: Dict[str, str]
    marriage_status: Optional[bool] = None
    weight: Annotated[float,Field(
            gt=0,
            title="Body Weight",
            description="Weight of the product in kilograms. Must be greater than 0.",
            examples=[1.5]
        )
    ]     #we will see annotations in .json 


# STEP 2: Define a function that accepts a Students object
def insert_student_data(student: Students):
    print(f"Name: {student.name}")
    print(f"Email: {student.email}")
    print(f"Age: {student.age}")
    print(f"LinkedIn: {student.LinkedIn}")
    print(f"Roll No: {student.roll_no}")
    print(f"CGPA: {student.cgpa}")
    print(f"Subjects: {student.subjects}")
    print(f"Contact Details: {student.contact_details}")
    print(f"Marriage Status: {student.marriage_status}")
    print(f"weight: {student.weight}")


# STEP 3: Create a dictionary containing raw student data
student1_info = {
    'name': 'Anisha Hazra',
    'email': 'anishahazra1201@gmail.com',
    'age': 21,
    'LinkedIn': 'https://www.linkedin.com/in/username/',
    'roll_no': 169,
    'cgpa': 7.8,
    'subjects': ['Maths', 'Science', 'Graphic Design'],
    'contact_details': {
        'city': 'Navi Mumbai',
        'Phone_number': '+91 9167395315'
    },
    'weight': 66
}


# STEP 4: Unpack the dictionary and create a Pydantic object
# Pydantic validates the data against the Students schema
student1 = Students(**student1_info)


# STEP 5: Pass the validated Students object to the function
insert_student_data(student1)
````
